## Introduction

This notebook builds a **portfolio slide deck** from the artifacts already generated in this repo.

Goal:
- Produce a clean, executive-friendly `.pptx` that explains:
  - the data (CT.gov at scale),
  - the scenario framing,
  - QUBO formulation,
  - classical baseline vs QAOA robustness,
  - AWS Braket SV1 results and decoding,
  - takeaways and next steps.

Outputs:
- `outputs/slides/quantum_clinical_trial_optimization_showcase.pptx`

Notes:
- It uses **existing result tables/figures** where possible.
- If an optional artifact is missing, it will skip that slide gracefully instead of failing.


```python
# ============================================================
# Cell 1 — Setup: imports, paths, and outputs
# ============================================================

from pathlib import Path
import pandas as pd
import json

# PPTX library
from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.enum.text import PP_ALIGN

RESULTS_DIR = Path("data/results")
FIG_DIR = Path("outputs/figures")
SLIDES_DIR = Path("outputs/slides")

FIG_DIR.mkdir(parents=True, exist_ok=True)
SLIDES_DIR.mkdir(parents=True, exist_ok=True)

OUT_PPTX = SLIDES_DIR / "quantum_clinical_trial_optimization_showcase.pptx"

print("Results:", RESULTS_DIR.resolve())
print("Figures:", FIG_DIR.resolve())
print("Slides :", SLIDES_DIR.resolve())
print("Will write:", OUT_PPTX.resolve())
```
